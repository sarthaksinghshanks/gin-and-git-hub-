```python
#block 7
# Load Cleaned Dataset 

df = pd.read_csv("spotify_cleaned.csv")

CHURN_COLUMN = None

possible_targets = [
    "churn",
    "churned",
    "is_churned",
    "customer_churn",
    "churn_status"
]

if CHURN_COLUMN is None:

    for column in possible_targets:

        if column in df.columns:

            CHURN_COLUMN = column
            break


if CHURN_COLUMN is None:

    raise ValueError(
        "Churn column not found. Set CHURN_COLUMN manually."
    )


print("\nSelected churn column:", CHURN_COLUMN)


df = df.dropna(subset=[CHURN_COLUMN]).copy()

target_values = (
    df[CHURN_COLUMN]
    .astype(str)
    .str.strip()
    .str.lower()
)

positive_values = [
    "1", "yes", "true", "churn", "churned"
]

negative_values = [
    "0", "no", "false", "not churn", "not churned", "retained"
]

unique_values = target_values.unique().tolist()
```

```python
#block 8
if set(unique_values).issubset(
    set(positive_values + negative_values)
):

    y = target_values.isin(positive_values).astype(int)

elif len(unique_values) == 2:

    print("Target categories:", unique_values)
    print("Second category is treated as positive.")

    y = (target_values == unique_values[1]).astype(int)

else:

    raise ValueError(
        "Churn target must have exactly two categories."
    )


if y.nunique() != 2:

    raise ValueError(
        "Both churn classes must be present in the dataset."
    )

# Prepare Input Features 
X = df.drop(columns=[CHURN_COLUMN])

id_columns = [
    column for column in X.columns
    if column.lower() in [
        "id", "user_id", "customer_id", "index"
    ]
]
```

```python
#block 9
X = X.drop(columns=id_columns)

numeric_columns = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

if len(numeric_columns) + len(categorical_columns) == 0:

    raise ValueError("No usable input features found.")


# Data Preprocessing 

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numerical", numeric_pipeline, numeric_columns),
    ("categorical", categorical_pipeline, categorical_columns)
])


# Create Logistic Regression Model 

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])
```

```python
#block 10
# Split Dataset 

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


# Train Model 

model.fit(X_train, y_train)

print("\nLogistic Regression model trained successfully!")



# Make Predictions

y_pred = model.predict(X_test)

y_probability = model.predict_proba(X_test)[:, 1]


# Evaluate Model 

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    y_probability
)

print("\nLOGISTIC REGRESSION RESULTS")
print("---------------------------")

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
print("ROC-AUC:", roc_auc)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
#block 11
# Saving Results

logistic_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC AUC"
    ],
    "Value": [
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    ]
})

logistic_results.to_csv(
    "logistic_results.csv",
    index=False
)

logistic_predictions = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Probability": y_probability
})

logistic_predictions.to_csv(
    "logistic_predictions.csv",
    index=False
)


# Confusion Matrix 

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()

plt.savefig("member2_confusion_matrix.png")
plt.show()
```

```python
#block 12
# ROC Curve 

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability
)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr,
    tpr,
    label="Logistic Regression"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")

plt.legend()
plt.tight_layout()

plt.savefig("member2_roc_curve.png")
plt.show()


# Precision-Recall Curve 

precision_values, recall_values, thresholds = (
    precision_recall_curve(y_test, y_probability)
)

plt.figure(figsize=(7, 5))

plt.plot(
    recall_values,
    precision_values
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")

plt.tight_layout()

plt.show()
```

```python
#block 13
# ============================================================
#              LINEAR REGRESSION
# ============================================================

# Load Cleaned Dataset 

df = pd.read_csv("spotify_cleaned.csv")
LINEAR_TARGET = None
possible_targets = [
    "listening_time",
    "listening_hours",
    "total_listening_time",
    "minutes_listened",
    "songs_played_per_day",
    "ads_listened_per_week"
]
if LINEAR_TARGET is None:

    for column in possible_targets:

        if column in df.columns:

            if pd.api.types.is_numeric_dtype(df[column]):

                LINEAR_TARGET = column
                break
if LINEAR_TARGET is None:

    raise ValueError(
        "Numerical target not detected. "
        "Set LINEAR_TARGET manually."
    )
print("\nSelected Linear Regression target:", LINEAR_TARGET)
df[LINEAR_TARGET] = pd.to_numeric(
    df[LINEAR_TARGET],
    errors="coerce"
)

df = df.dropna(subset=[LINEAR_TARGET]).copy()

X = df.drop(columns=[LINEAR_TARGET])

y = df[LINEAR_TARGET]

id_columns = [
    column for column in X.columns
    if column.lower() in [
        "id", "user_id", "customer_id", "index"
    ]
]

X = X.drop(columns=id_columns)

numeric_columns = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

if len(numeric_columns) + len(categorical_columns) == 0:

    raise ValueError("No usable input features found.")
```

```python
#block 14
# Preprocessing Data

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numerical", numeric_pipeline, numeric_columns),
    ("categorical", categorical_pipeline, categorical_columns)
])


# Create Linear Regression Model 

model = Pipeline([
    ("preprocessor", preprocessor),
    ("regression", LinearRegression())
])
```

```python
#block 15
# Split Dataset 

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


# Training Model For Linear Regressinon  

model.fit(X_train, y_train)

print("\nLinear Regression model trained successfully!")


# Make Predictions 

y_pred = model.predict(X_test)


# Evaluating Model 

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("\nLINEAR REGRESSION RESULTS")
print("-------------------------")

print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
print("R2 Score:", r2)
```

```python
#block 16
# Saving Results

linear_results = pd.DataFrame({
    "Metric": [
        "MAE",
        "MSE",
        "RMSE",
        "R2 Score"
    ],
    "Value": [
        mae,
        mse,
        rmse,
        r2
    ]
})

linear_results.to_csv(
    "linear_results.csv",
    index=False
)

linear_predictions = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Residual": y_test.values - y_pred
})

linear_predictions.to_csv(
    "linear_predictions.csv",
    index=False
)
```
